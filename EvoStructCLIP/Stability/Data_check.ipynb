```python
import pandas as pd

# 1. S669 데이터 로드
file_path = r'C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\Stability\S669.tsv'
df = pd.read_csv(file_path, sep='\t')

# 2. S669에 포함된 고유 UniProt ID 추출
s669_ids = set(df['PDB'].unique())

# 3. 현재 가지고 있는 msa_dict의 키(ID) 추출
msa_keys = msa_dict.keys()

# 4. 비교 분석
existing_msa = s669_ids.intersection(msa_keys)
missing_msa = s669_ids - msa_keys

print(f"전체 S669 단백질 수: {len(s669_ids)}")
print(f"이미 MSA가 있는 단백질 수: {len(existing_msa)}")
print(f"MSA를 새로 구축해야 할 단백질 수: {len(missing_msa)}")

missing_msa
```

```python
import os

# 1. 경로 설정
source_fasta = r'C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\Stability\S669.fasta'
output_dir = r'C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\Stability\S669_fasta_files'

# 2. 출력 폴더 생성
if not os.path.exists(output_dir):
    os.makedirs(output_dir)
    print(f"폴더 생성 완료: {output_dir}")

# 3. 파일 읽기 및 분리 저장
with open(source_fasta, 'r') as f:
    content = f.read()
    # '>'를 기준으로 분할 (첫 번째 빈 결과는 제외)
    entries = [('>' + e).strip() for e in content.split('>') if e.strip()]

for entry in entries:
    # 첫 번째 줄에서 ID 추출 (공백이나 특수문자 제거)
    lines = entry.split('\n')
    header = lines[0]
    seq_id = header[1:].split()[0] # '>' 제외하고 첫 단어만 ID로 사용
    
    # 파일명 규칙: ID.fasta
    file_name = f"{seq_id}.fasta"
    full_path = os.path.join(output_dir, file_name)
    
    with open(full_path, 'w') as out_f:
        out_f.write(entry)

print(f"총 {len(entries)}개의 FASTA 파일이 분리 저장되었습니다.")
```
