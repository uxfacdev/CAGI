```python
import pandas as pd
import os
from Bio.PDB import PDBParser

# 3문자 아미노산 코드를 1문자로 변환하는 사전
d3to1 = {
    'ALA': 'A', 'CYS': 'C', 'ASP': 'D', 'GLU': 'E', 'PHE': 'F',
    'GLY': 'G', 'HIS': 'H', 'ILE': 'I', 'LYS': 'K', 'LEU': 'L',
    'MET': 'M', 'ASN': 'N', 'PRO': 'P', 'GLN': 'Q', 'ARG': 'R',
    'SER': 'S', 'THR': 'T', 'VAL': 'V', 'TRP': 'W', 'TYR': 'Y'
}

def verify_pdb_residues(df, pdb_dir):
    parser = PDBParser(QUIET=True)
    check_results = []
    actual_res_list = []

    print("--- PDB WT 잔기 체크 시작 ---")
    
    for idx, row in df.iterrows():
        pdb_id = row['PDB_ID']
        chain_id = row['Chain']
        # PDB 포지션은 정수형으로 변환 (간혹 insertion code가 있는 경우 예외처리 필요)
        try:
            pos = int(row['MutPos'])
        except ValueError:
            check_results.append("Invalid Position Format")
            actual_res_list.append("N/A")
            continue

        expected_wt = row['WT']
        pdb_path = os.path.join(pdb_dir, f"{pdb_id}.pdb")

        # 1. 파일 존재 여부 확인
        if not os.path.exists(pdb_path):
            check_results.append("File Missing")
            actual_res_list.append("N/A")
            continue

        try:
            # 2. PDB 파싱
            structure = parser.get_structure(pdb_id, pdb_path)
            model = structure[0]
            
            # 3. 체인 및 잔기 접근
            if chain_id not in model:
                check_results.append(f"Chain {chain_id} Missing")
                actual_res_list.append("N/A")
                continue
                
            chain = model[chain_id]
            
            # PDB 잔기 ID 형식: ('hetero-flag', seq-number, 'insertion-code')
            # 일반적인 아미노산은 (' ', pos, ' ') 형태입니다.
            res_id = (' ', pos, ' ')
            
            if res_id in chain:
                residue = chain[res_id]
                res_name_3 = residue.get_resname()
                res_name_1 = d3to1.get(res_name_3, 'X') # 매핑 안되면 X
                
                actual_res_list.append(res_name_1)
                if res_name_1 == expected_wt:
                    check_results.append("Match")
                else:
                    check_results.append(f"Mismatch (Found: {res_name_1})")
            else:
                check_results.append("Position Missing in PDB")
                actual_res_list.append("N/A")

        except Exception as e:
            check_results.append(f"Error: {str(e)}")
            actual_res_list.append("N/A")

        if (idx + 1) % 50 == 0:
            print(f"진행 중... [{idx + 1}/{len(df)}]")

    # 결과 기록
    df['Actual_WT'] = actual_res_list
    df['Verify_Result'] = check_results
    return df

# --- 실행 부분 ---
# 사용자의 실제 경로 설정 (r을 붙여 raw string으로 처리)
pdb_dir_path = r"C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\S669\S669_Zenodo\pdbs"

# df_zenodo는 이미 로드되어 있다고 가정합니다.
df_verified = verify_pdb_residues(df_zenodo, pdb_dir_path)

# 결과 요약 출력
print("\n[ 검증 결과 요약 ]")
print(df_verified['Verify_Result'].value_counts())

# 불일치 항목만 따로 보기
mismatches = df_verified[df_verified['Verify_Result'].str.contains("Mismatch|Missing")]
if not mismatches.empty:
    print("\n[ 주의: 불일치 또는 누락된 항목 ]")
    print(mismatches[['PDB_ID', 'Chain', 'MutPos', 'WT', 'Actual_WT', 'Verify_Result']])
```

```python
import pandas as pd
import os
from Bio.PDB import PDBParser, PPBuilder

def generate_s669_fasta(df, pdb_dir, output_file):
    parser = PDBParser(QUIET=True)
    ppb = PPBuilder()
    
    # 중복된 PDB_ID와 Chain 조합 제거 (서열은 체인당 하나면 되므로)
    unique_targets = df[['PDB_ID', 'Chain']].drop_duplicates()
    
    print(f"--- FASTA 생성 시작 (대상: {len(unique_targets)}개 체인) ---")
    
    with open(output_file, 'w') as f:
        count = 0
        for _, row in unique_targets.iterrows():
            pdb_id = row['PDB_ID']
            chain_id = row['Chain']
            pdb_path = os.path.join(pdb_dir, f"{pdb_id}.pdb")
            
            if not os.path.exists(pdb_path):
                print(f"Warning: {pdb_id}.pdb 파일을 찾을 수 없습니다.")
                continue
                
            try:
                structure = parser.get_structure(pdb_id, pdb_path)
                model = structure[0]
                
                if chain_id in model:
                    # PPBuilder는 ATOM 레코드에서 연속된 폴리펩타이드를 찾아 서열로 변환합니다.
                    # 중간에 끊긴 루프가 있을 경우 여러 개의 조각(pp)이 나올 수 있어 합쳐줍니다.
                    chain_seq = ""
                    for pp in ppb.build_peptides(model[chain_id]):
                        chain_seq += str(pp.get_sequence())
                    
                    if chain_seq:
                        # 헤더 형식: >PDBID_Chain
                        f.write(f">{pdb_id}_{chain_id}\n")
                        f.write(f"{chain_seq}\n")
                        count += 1
                else:
                    print(f"Warning: {pdb_id}.pdb 내에 {chain_id} 체인이 없습니다.")
                    
            except Exception as e:
                print(f"Error processing {pdb_id}: {str(e)}")

    print(f"--- 완료! 총 {count}개의 서열이 {output_file}에 저장되었습니다. ---")

# --- 실행 부분 ---
df_zenodo = pd.read_csv(r"C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\S669\S669_zenodo.tsv", sep ="\t")
pdb_dir_path = r"C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\S669\S669_Zenodo\pdbs"
output_fasta = "S669_zenodo.fasta"

# 이전에 검증에 사용했던 df_zenodo를 그대로 사용합니다.
generate_s669_fasta(df_zenodo, pdb_dir_path, output_fasta)
```

```python
import pandas as pd
import os
from Bio.PDB import PDBParser, PPBuilder

def generate_s669_fasta_with_pos(df, pdb_dir, output_file):
    parser = PDBParser(QUIET=True)
    ppb = PPBuilder()
    
    # 3문자 -> 1문자 변환용 (PPBuilder 내부에서 쓰이지만 매핑 확인용으로 선언)
    d3to1 = {'ALA': 'A', 'CYS': 'C', 'ASP': 'D', 'GLU': 'E', 'PHE': 'F', 'GLY': 'G', 'HIS': 'H', 
             'ILE': 'I', 'LYS': 'K', 'LEU': 'L', 'MET': 'M', 'ASN': 'N', 'PRO': 'P', 'GLN': 'Q', 
             'ARG': 'R', 'SER': 'S', 'THR': 'T', 'VAL': 'V', 'TRP': 'W', 'TYR': 'Y'}

    # 결과를 저장할 리스트 (나중에 df에 합침)
    seq_pos_list = [None] * len(df)
    
    # 유니크한 PDB-Chain 조합별로 처리
    unique_targets = df[['PDB_ID', 'Chain']].drop_duplicates()
    
    print(f"--- FASTA 생성 및 포지션 매핑 시작 ---")
    
    with open(output_file, 'w') as f_out:
        for _, target in unique_targets.iterrows():
            pdb_id, chain_id = target['PDB_ID'], target['Chain']
            pdb_path = os.path.join(pdb_dir, f"{pdb_id}.pdb")
            
            if not os.path.exists(pdb_path):
                continue
                
            try:
                structure = parser.get_structure(pdb_id, pdb_path)
                model = structure[0]
                if chain_id not in model: continue
                
                # 1. PPBuilder로 물리적으로 연결된 서열 조각(Polypeptides) 추출
                polypeptides = ppb.build_peptides(model[chain_id])
                
                full_seq = ""
                # FASTA 내 각 글자가 실제 PDB에서 몇 번이었는지 기록할 리스트
                pdb_res_nums = [] 

                for pp in polypeptides:
                    full_seq += str(pp.get_sequence())
                    # 각 잔기의 PDB 번호(id[1])를 순서대로 추출
                    for residue in pp:
                        pdb_res_nums.append(residue.get_id()[1])
                
                if not full_seq: continue

                # 2. FASTA 파일 쓰기
                f_out.write(f">{pdb_id}_{chain_id}\n{full_seq}\n")

                # 3. 해당 PDB_ID/Chain을 쓰는 모든 행(Mutation)에 대해 SeqPos 계산
                row_indices = df[(df['PDB_ID'] == pdb_id) & (df['Chain'] == chain_id)].index
                
                for idx in row_indices:
                    target_pdb_pos = int(df.at[idx, 'MutPos'])
                    
                    # PDB 번호 리스트에서 해당 번호가 몇 번째(index)에 있는지 찾기
                    try:
                        # 0-based index에 1을 더해 1-based position 생성
                        seq_index = pdb_res_nums.index(target_pdb_pos)
                        seq_pos_list[idx] = seq_index + 1 
                    except ValueError:
                        # PDB 구조에 좌표가 없는 번호인 경우
                        seq_pos_list[idx] = None
                        
            except Exception as e:
                print(f"Error in {pdb_id}_{chain_id}: {e}")

    # 4. 데이터프레임 업데이트
    df['SeqPos'] = seq_pos_list
    print(f"--- 완료! FASTA가 저장되었고 df_zenodo에 'SeqPos' 컬럼이 추가되었습니다. ---")
    return df

# --- 실행 부분 ---
df_zenodo = pd.read_csv(r"C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\S669\S669_zenodo.tsv", sep ="\t")
pdb_dir_path = r"C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\S669\S669_Zenodo\pdbs"
output_fasta = "S669_zenodo.fasta"

# 함수 실행 및 결과 업데이트
df_zenodo = generate_s669_fasta_with_pos(df_zenodo, pdb_dir_path, output_fasta)

# 결과 확인
print(df_zenodo[['PDB_ID', 'Chain', 'MutPos', 'SeqPos', 'WT']].head())
```

```python
import pandas as pd

def analyze_pos_differences(df):
    # 1. 오프셋 계산 (PDB번호 - FASTA순서)
    # SeqPos가 NaN인 경우(구조에 좌표가 없는 경우)는 제외하고 계산합니다.
    df_valid = df.dropna(subset=['SeqPos']).copy()
    df_valid['Offset'] = df_valid['MutPos'].astype(int) - df_valid['SeqPos'].astype(int)

    # 2. Key 단위로 그룹화하여 통계 산출
    key_stats = df_valid.groupby('Key')['Offset'].agg(['min', 'max', 'mean', 'nunique', 'std']).reset_index()
    
    # 3. 결과 분석 및 카운팅
    # 오프셋이 일정한 그룹 (std == 0 또는 nunique == 1)
    consistent_keys = key_stats[key_stats['nunique'] == 1]
    
    # 오프셋이 상이한 그룹 (중간에 Gap이 있거나 구조가 끊긴 경우)
    inconsistent_keys = key_stats[key_stats['nunique'] > 1]
    
    # [추가] 오프셋이 0이 아닌 케이스 계산
    # 1) 일정한데 0이 아닌 경우 (단순 오프셋 존재)
    nonzero_consistent = consistent_keys[consistent_keys['min'] != 0]
    # 2) 일정하지 않은 경우 (Gap 존재 등, 당연히 0이 아님)
    nonzero_total_count = len(nonzero_consistent) + len(inconsistent_keys)
    # 3) 완벽하게 1:1 매칭되는 경우 (Offset이 0인 경우)
    perfect_match_count = len(consistent_keys[consistent_keys['min'] == 0])

    print(f" 전체 요약 ")
    print(f"- 분석 대상 Key 개수: {len(key_stats)}")
    print(f"- 오프셋이 0인 Key (완벽 일치): {perfect_match_count}")
    print(f"- 오프셋이 0이 아닌 Key (오프셋/Gap 존재): {nonzero_total_count}") # 요청하신 부분
    print(f"  └ 오프셋이 일정하나 0은 아님: {len(nonzero_consistent)}")
    print(f"  └ 오프셋이 불규칙함 (Gap 의심): {len(inconsistent_keys)}")
    print("-" * 30)

    if not inconsistent_keys.empty:
        print("\n[ 오프셋이 상이한 Key 리스트 (주의 필요) ]")
        print(inconsistent_keys.sort_values('std', ascending=False).head(10))
    
    if not nonzero_consistent.empty:
        print("\n[ 오프셋이 일정하지만 0은 아닌 Key 샘플 ]")
        print(nonzero_consistent[['Key', 'min']].rename(columns={'min': 'Constant_Offset'}).head(10))

    return key_stats, inconsistent_keys

# 실행 예시
df_zenodo = pd.read_csv(r"C:\Users\Kunny\Documents\GitHub\CAGI\EvoStructCLIP\S669\S669_zenodo.tsv", sep="\t")
key_summary, gap_keys = analyze_pos_differences(df_zenodo)
```
