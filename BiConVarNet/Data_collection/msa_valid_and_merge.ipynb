```python
import pickle

# 파일 경로 설정
pkl_path_old = r"E:\CAGI_data\msa_dict_valid.pkl"
pkl_path_new = r"E:\CAGI_data\msa_dict_valid_new.pkl"

# pickle 파일 로드
with open(pkl_path_old, "rb") as f:
    msa_old = pickle.load(f)

with open(pkl_path_new, "rb") as f:
    msa_new = pickle.load(f)

# 비교
common_uids = set(msa_old.keys()) & set(msa_new.keys())
different = []
missing_in_old = []
missing_in_new = []

for uid in common_uids:
    if msa_old[uid] != msa_new[uid]:
        different.append(uid)

for uid in msa_old:
    if uid not in msa_new:
        missing_in_new.append(uid)

for uid in msa_new:
    if uid not in msa_old:
        missing_in_old.append(uid)

# 결과 출력
print(f"총 uid 수: old={len(msa_old)}, new={len(msa_new)}")
print(f"공통 uid 수: {len(common_uids)}")
print(f"서로 다른 uid 수: {len(different)}")
if different:
    print("차이나는 예시:")
    for uid in different[:3]:
        print(f"\n[UID: {uid}]")
        print("OLD:", msa_old[uid][:2])
        print("NEW:", msa_new[uid][:2])

if missing_in_new:
    print(f"new에 없는데 old에만 있는 uid 수: {len(missing_in_new)}")

if missing_in_old:
    print(f"old에 없는데 new에만 있는 uid 수: {len(missing_in_old)}")
```
