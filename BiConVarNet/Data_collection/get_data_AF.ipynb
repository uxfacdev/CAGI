```python
import os
from tqdm import tqdm
import pandas as pd
import requests

# TSV 파일 경로
tsv_path = "rhapsody2_sav_db.tsv"

# 저장 경로
save_dir = r"C:\Users\Kunny\Research\Dataset\Missense Variant dataset\alphafold_structures"
os.makedirs(save_dir, exist_ok=True)

# 실패 로그 파일 경로
log_path = os.path.join(save_dir, "failed_log.txt")

# 구조 파일 이름에서 다운로드 URL을 생성
def download_af_pdb(structure_file, save_dir):
    url = f"https://alphafold.ebi.ac.uk/files/{structure_file}"
    save_path = os.path.join(save_dir, structure_file)
    if not os.path.exists(save_path):
        r = requests.get(url)
        if r.ok:
            with open(save_path, "w") as f:
                f.write(r.text)
            return True
        else:
            return False
    return True

# 중복 제거된 structure 파일 기준 다운로드
structure_files = df["StructureFile"].unique()
failed_files = []

for pdb_name in tqdm(structure_files):
    success = download_af_pdb(pdb_name, save_dir)
    if not success:
        print(f"❌ Failed: {pdb_name}")
        failed_files.append(pdb_name)

# 실패 파일 로그 저장
if failed_files:
    with open(log_path, "w") as f:
        for pdb in failed_files:
            f.write(f"{pdb}\n")
    print(f"\n📄 Failed downloads saved to: {log_path}")
else:
    print("\n✅ All files downloaded successfully.")
```

```python
import os
import tarfile
from tqdm import tqdm
import shutil
import gzip

# 경로 설정
tar_path = r"C:\Users\Kunny\Downloads\UP000005640_9606_HUMAN_v4.tar"
save_dir = r"C:\Users\Kunny\Research\Dataset\Missense Variant dataset\alphafold_structures"
log_path = os.path.join(save_dir, "failed_log.txt")

# 실패 파일 목록 불러오기
with open(log_path, "r") as f:
    failed_files = [line.strip() for line in f if line.strip()]

# .pdb → .pdb.gz 매핑용 dict
gz_map = {f"{name}.gz": name for name in failed_files}

# 결과 기록용
still_failed = []

# .tar 파일 열기
with tarfile.open(tar_path, "r") as tar:
    members = tar.getmembers()
    member_names = {os.path.basename(m.name): m for m in members}

    for gz_name, pdb_name in tqdm(gz_map.items()):
        if gz_name in member_names:
            member = member_names[gz_name]
            try:
                extracted_file = tar.extractfile(member)
                if extracted_file is None:
                    raise Exception("File not found in archive")

                # 압축 해제하여 .pdb로 저장
                out_path = os.path.join(save_dir, pdb_name)
                with gzip.open(extracted_file, "rb") as gz_f:
                    with open(out_path, "wb") as out_f:
                        shutil.copyfileobj(gz_f, out_f)
            except Exception as e:
                print(f"❌ Error decompressing {gz_name}: {e}")
                still_failed.append(pdb_name)
        else:
            still_failed.append(pdb_name)

# 실패 로그 업데이트
with open(log_path, "w") as f:
    for name in still_failed:
        f.write(name + "\n")

# 요약 출력
print(f"\n✅ 성공적으로 복구된 파일 수: {len(failed_files) - len(still_failed)}개")
print(f"❌ 여전히 실패한 파일 수: {len(still_failed)}개 → {log_path}에 저장됨")
```
