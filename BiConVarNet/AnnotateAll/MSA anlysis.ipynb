```python
import numpy as np
import matplotlib.pyplot as plt
import pickle

with open(r"E:\CAGI_data\msa_dict_valid.pkl", "rb") as f:
    msa_dict = pickle.load(f)

# msa_dict: {query_id: [(name, seq), ...]} 구조로 되어 있어야 함
msa_depths = {qid: len(seq_list) for qid, seq_list in msa_dict.items()}
depth_values = np.array(list(msa_depths.values()))

# 기본 통계 출력
print(f"Total Queries: {len(depth_values)}")
print(f"Average MSA Depth: {np.mean(depth_values):.2f}")
print(f"Median MSA Depth: {np.median(depth_values):.2f}")
print(f"Min MSA Depth: {np.min(depth_values)}")
print(f"Max MSA Depth: {np.max(depth_values)}")
print(f"<= 1 Sequences: {(depth_values <= 1).sum()} queries")
print(f"<= 10 Sequences: {(depth_values <= 10).sum()} queries")
print(f"<= 30 Sequences: {(depth_values <= 30).sum()} queries")
print(f"<= 80 Sequences: {(depth_values <= 80).sum()} queries")
print(f"<= 100 Sequences: {(depth_values <= 100).sum()} queries")
print(f">= 1000 Sequences: {(depth_values >= 1000).sum()} queries")

# 히스토그램 출력 (log scale)
plt.figure(figsize=(10, 5))
plt.hist(depth_values, bins=50, log=True, color='skyblue', edgecolor='black')
plt.title("Histogram of MSA Depths per Query")
plt.xlabel("MSA Depth (# sequences per query)")
plt.ylabel("Number of Queries (log scale)")
plt.grid(True)
plt.show()
```
